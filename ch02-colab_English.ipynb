## My First Deep Learning Model: A Preview of the Code from Chapter 10, 'Predicting Survival Rates for Lung Cancer Surgery Patients'

### 1. Setting Up the Environment 

```python
from tensorflow.keras.models import Sequential  # Import the functions needed from TensorFlow's Keras API.
from tensorflow.keras.layers import Dense       # Import the library needed to work with data.
import numpy as np
```

### 2. Preparing the Data

```python
!git clone https://github.com/taehojo/data.git   # Download the data provided on GitHub.

Data_set = np.loadtxt("./data/ThoraricSurgery3.csv", delimiter=",")  # Load the surgical patient data.
X = Data_set[:,0:16]                                                 # Assign the patients' examination records to X.
y = Data_set[:,16]                                                   # Assign the postoperative death/survival outcomes to y.
```

### 3. Defining the Model Architecture 

```python
model = Sequential()                                                  # Define the architecture of the deep learning model.
model.add(Dense(30, input_dim=16, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

### 4. Running the Model 

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])  # Run the deep learning model.
history=model.fit(X, y, epochs=5, batch_size=16)
```
