# Chapter 10. Designing a Deep Learning Model (Chapter 2: A Preview of Deep Learning Essentials)

### 1. Setting Up the Environment

```python
# Import the required functions from the Keras API in TensorFlow.
from tensorflow.keras.models import Sequential  
from tensorflow.keras.layers import Dense   

# Import the library needed to work with data.
import numpy as np
```

### 2. Preparing the Data

```python
# Download the data provided on GitHub.
!git clone https://github.com/taehojo/data.git   

# Load the prepared surgical patient data.
Data_set = np.loadtxt("./data/ThoraricSurgery3.csv", delimiter=",")  
X = Data_set[:,0:16]    # Assign the patients' examination records to X.
y = Data_set[:,16]      # Assign death/survival status one year after surgery to y.
```

### 3. Defining the Architecture

```python
# Define the architecture of the deep learning model.
model = Sequential()                                                   
model.add(Dense(30, input_dim=16, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

### 4. Running the Model

```python
# Run the deep learning model.
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy']) 
history=model.fit(X, y, epochs=5, batch_size=16)
```
