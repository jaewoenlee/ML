### 1. Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import SGD
```

### 2. Prepare OR gate data

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 1])
```

### 3. Build the model

```python
model = Sequential()
model.add(Input(shape=(2,)))
model.add(Dense(1, activation='sigmoid',
                kernel_initializer='zeros', bias_initializer='zeros'))
```

### 4. Train the model

```python
model.compile(loss='mse', optimizer=SGD(learning_rate=0.1))
history = model.fit(X, y, epochs=1000, batch_size=4, shuffle=False, verbose=0)

for epoch in range(0, 1000, 100):
    print('epoch =', epoch, 'MSE =', history.history['loss'][epoch])
```

### 5. Show the learned weights and OR truth table

```python
weights, bias = model.layers[0].get_weights()
w1, w2 = weights[:, 0]
b = bias[0]
print('w1 =', w1, 'w2 =', w2, 'b =', b)

pred = model.predict(X, verbose=0).flatten()
print('x1 x2 | y | y_pred | result')
for (x1, x2), target, p in zip(X, y, pred):
    print(x1, x2, '|', target, '|', round(float(p), 4), '|', int(p >= 0.5))
```

### 6. Plot the decision boundary

```python
plt.scatter(X[y == 0, 0], X[y == 0, 1], s=120, facecolors='white', edgecolors='black')
plt.scatter(X[y == 1, 0], X[y == 1, 1], s=120, facecolors='black', edgecolors='black')

x1_line = np.array([-0.5, 1.5])
plt.plot(x1_line, -(w1 * x1_line + b) / w2)
plt.xlabel('x1')
plt.ylabel('x2')
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.grid(True)
plt.show()
```
