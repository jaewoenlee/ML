# Chapter 10: Designing Deep Learning Models (Preview of Core Deep Learning Concepts from Chapter 2) — Prof. Taeho Jo

### 1. Setting Up the Environment

```python
# Import the required classes from the Keras API in TensorFlow.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Import NumPy for working with data.
import numpy as np
```

### 2. Preparing the Data

```python
# Clone the data repository from GitHub.
!git clone https://github.com/taehojo/data.git

# Load the provided thoracic surgery patient data.
Data_set = np.loadtxt("./data/ThoraricSurgery3.csv", delimiter=",")
X = Data_set[:,0:16]    # Assign patients' examination records to X.
y = Data_set[:,16]      # Assign one-year postoperative mortality/survival outcomes to y.
```

### 3. Defining the Model Architecture

```python
# Define the deep learning model architecture.
model = Sequential()
model.add(Dense(30, input_dim=16, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

### 4. Training the Model

```python
# Compile and train the deep learning model.
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
history=model.fit(X, y, epochs=5, batch_size=16)
```

### 5. Making Predictions on the First 20 Samples

```python
# Select the first 20 samples.
X_sample_20 = X[:20]
y_sample_20 = y[:20]

# Generate predictions using the model.
predictions_20 = model.predict(X_sample_20)

# Display the results.
print("Actual vs. predicted values (first 20 samples):")
print("-" * 55)
print(f"{'No.':^6} | {'Actual Survival Status (y)':^15} | {'Predicted Probability (Model Output)':^22} | {'Predicted Class':^10}")
print("-" * 55)
for i in range(20):
    true_val = int(y_sample_20[i])
    pred_prob = predictions_20[i][0]
    pred_val = 1 if pred_prob >= 0.5 else 0
    print(f"{i+1:^8} | {true_val:^17} | {pred_prob:^24.4f} | {pred_val:^10}")
print("-" * 55)
```
