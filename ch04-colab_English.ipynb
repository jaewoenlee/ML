## 4. Exploring Least Squares with Python

### 1. Setting Up the Environment

```python
import numpy as np
```

### 2. Preparing the Data

```python
# Create NumPy arrays named x and y for study hours and scores, respectively.
x = np.array([2, 4, 6, 8])
y = np.array([81, 93, 91, 97]) 
```

#### [Mean Values of x and y]

```python
# Calculate the mean of x.
mx = np.mean(x)

# Calculate the mean of y.
my = np.mean(y)

# Print the results to check them.
print("x의 평균값:", mx)
print("y의 평균값:", my)
```

#### [Denominator and Numerator of the Slope Formula]

```python
# This is the denominator of the slope formula.
divisor = sum([(i - mx)**2 for i in x])

# This is the numerator of the slope formula.
def top(x, mx, y, my):
    d = 0
    for i in range(len(x)):
        d += (x[i] - mx) * (y[i] - my)
    return d
dividend = top(x, mx, y, my)

# Print the results to check them.
print("분모:", divisor)
print("분자:", dividend)
```

### 3. Calculating the Slope and y-Intercept

```python
# Use this formula to calculate the slope a.
a = dividend / divisor

# Use this formula to calculate the y-intercept b.
b = my - (mx*a)

# Print the results to check them.
print("기울기 a =", a)
print("y절편 b =", b)
```

## 6. Exploring Mean Squared Error with Python

### 1. Setting Up the Environment

```python
import numpy as np
```

### 2. Preparing the Data

```python
# Choose hypothetical values for the slope a and y-intercept b.
fake_a=3
fake_b=76

# Create NumPy arrays for study hours x and scores y.
x = np.array([2, 4, 6, 8])
y = np.array([81, 93, 91, 97]) 
```

### 3. Calculating Mean Squared Error

```python
# This function returns the result of substituting hypothetical a,b values into y=ax + b.
def predict(x):
    return fake_a * x + fake_b

# Create an empty list to hold the predictions.
predict_result = []

# Substitute each x value to populate the predict_result list.
for i in range(len(x)):
    predict_result.append(predict(x[i]))
    print("공부시간=%.f, 실제점수=%.f, 예측점수=%.f" % (x[i], y[i], predict(x[i])))
```

```python
# This function applies the mean squared error function to the y values to obtain the final result.
n=len(x)  
def mse(y, y_pred):
    return (1/n) * sum((y - y_pred)**2)

# Print the mean squared error.
print("평균 제곱 오차: " + str(mse(y,predict_result)))
```
