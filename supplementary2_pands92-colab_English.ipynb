## M. Merging Data

```python
# Create DataFrame 1 for this exercise.
adf = pd.DataFrame({"x1" : ["A","B","C"], 
                    "x2": [1,2,3]})
adf
```

```python
# Create DataFrame 2.
bdf = pd.DataFrame({"x1" : ["A","B","D"], 
                    "x3": ["T","F","T"]})
bdf
```

```python
# Create DataFrame 3.
cdf = pd.DataFrame({"x1" : ["B","C","D"], 
                    "x2": [2,3,4]})
cdf
```

```python
# 70. Merge using the left-hand columns as the reference.
pd.merge(adf, bdf, how='left', on='x1')   
# Merge on x1, using the left DataFrame (adf) as the reference.
# Since the left DataFrame adf has no D, the corresponding values become NaN.
```

```python
# 71. Merge using the right-hand columns as the reference.
pd.merge(adf, bdf, how='right', on='x1')   
# Merge on x1, using the right DataFrame (bdf) as the reference.
# Since the right DataFrame bdf has no C, the corresponding values become NaN.
```

```python
# 72. Merge only shared values.
pd.merge(adf, bdf, how='inner', on='x1')
```

```python
# 73. Merge all values.
pd.merge(adf, bdf, how='outer', on='x1')
```

```python
# 74. Compare specific columns and retrieve only records with shared values.
adf[adf.x1.isin(bdf.x1)]  
# Compare specific columns in adf and bdf, retrieving only records with shared values.
# adf.x1 and bdf.x1 share A,B. Display only the corresponding records from adf.
```

```python
# 75. Exclude records with shared values when merging.
adf[~adf.x1.isin(bdf.x1)]  # Display everything except those records.
```

```python
# 76. Merge only records with shared values.
pd.merge(adf, cdf)
```

```python
# 77. Merge everything.
pd.merge(adf, cdf, how='outer')
```

```python
# 78. Indicate where each merged record came from.
pd.merge(adf, cdf, how='outer', indicator=True)
```

```python
# 79. Keep only the desired merge results.
pd.merge(adf, cdf, how='outer', indicator=True).query('_merge == "left_only"')
```

```python
# 80. Remove the merge column.
pd.merge(adf, cdf, how='outer', indicator=True).query('_merge == "left_only"').drop(columns=['_merge'])
```
